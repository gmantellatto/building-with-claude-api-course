```python
# Load env variables
from dotenv import load_dotenv

load_dotenv()
```

```python
# Create an API client
from anthropic import Anthropic
import os

token = os.environ["ANTHROPIC_AUTH_TOKEN"]
url = os.environ["ANTHROPIC_BASE_URL"]
model = os.environ["ANTHROPIC_MODEL"]

client = Anthropic(
    api_key=token,
    base_url=url,
    default_headers={"Authorization": f"Bearer {token}"}
)
```

```python
# Make a request

message = client.messages.create(
  model=model,
  max_tokens=1000,
  messages=[
    {
      "role": "user",
      "content": "What is quantum computing? Answer in one sentence."
    }
  ]
)
```

```python
# Print the response

print(message)

# Print just the generated text

print(message.content[0].text)
```

```python
# Messages send to the Claude API do not relate to each other.

message = client.messages.create(
  model=model,
  max_tokens=1000,
  messages=[
    {
      "role": "user",
      "content": "Write another sentence."
    }
  ]
)

# The output will not relate to the previous message, as each message is independent.
print(message.content[0].text)
```

```python
# To solve this, you can send a list of messages to the API, which will allow the model to maintain context across multiple messages.

# For this, we can create 3 helper functions to add messages to a list and send them to the API.

def add_user_message(messages, content):
    user_message = { "role": "user", "content": content }
    messages.append(user_message)

def add_assistant_message(messages, content):
    assistant_message = { "role": "assistant", "content": content }
    messages.append(assistant_message)

def chat(messages):
  message = client.messages.create(
    model=model,
    max_tokens=1000,
    messages= messages
  )
  return message.content[0].text
```

```python
# Make a starting list of messages
messages = []

# Add in the initial user question
add_user_message(messages, "What is quantum computing? Answer in one sentence.")

# Pass the list of messages into 'chat' to get an answer
answer = chat(messages)

# Take the answer and add it as an assistant message into our list
add_assistant_message(messages, answer)

# Add in the user's follow-up question
add_user_message(messages, "Write another sentence.")

# Call chat again with the list of messages to get a final answer
answer = chat(messages)
add_assistant_message(messages, answer)

answer
```
