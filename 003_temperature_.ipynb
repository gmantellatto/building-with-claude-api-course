```python
# Create an API client and Helper Functions
from anthropic import Anthropic
from dotenv import load_dotenv
import os

load_dotenv()

token = os.environ["ANTHROPIC_AUTH_TOKEN"]
url = os.environ["ANTHROPIC_BASE_URL"]
model = os.environ["ANTHROPIC_MODEL"]

client = Anthropic(
    api_key=token,
    base_url=url,
    default_headers={"Authorization": f"Bearer {token}"}
)

def add_user_message(messages, content):
    user_message = { "role": "user", "content": content }
    messages.append(user_message)

def add_assistant_message(messages, content):
    assistant_message = { "role": "assistant", "content": content }
    messages.append(assistant_message)
```

```python
# Updating the chat function to accept a temperature parameter
def chat(messages, system=None, temperature=1.0):

  params = {
    "model": model,
    "max_tokens": 1000,
    "messages": messages,
    "temperature": temperature,
  }

  if system:
    params["system"] = system
  
  message = client.messages.create(**params)
  return message.content[0].text
```

```python
# The higher the temperature, the more random (creative) the output. The lower the temperature, the more deterministic the output. 
# A temperature of 0.0 will always produce the same output for a given input.

messages = []
add_user_message(
  messages,
  "Generate a one sentence movie idea"
)
answer = chat(messages, temperature=0.0)
print("Temperature 0.0:", answer)
```

```python
messages = []
add_user_message(
  messages,
  "Generate a one sentence movie idea"
)
answer = chat(messages, temperature=1.0)
print("Temperature 1.0:", answer)
```
